```python
# Start with imports - ask ChatGPT to explain any package that you don't know

import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
request = "What is remainder theorem in Polynomials? Give me some examples for understanding it better. "
request += "Answer with relevant examples and in a way that a 15 year old can understand it. "
messages = [{"role": "user", "content": request}]
```

```python
messages
```

```python
# Run the model in stream mode with max tokens limit
ollama = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
model_name = "deepseek-r1:8B"

# Create a streaming response
print("Generating response in real-time (limited to 1500 tokens)...\n")
full_response = ""

# Add instructions for LaTeX formatting to the messages
latex_instruction = "Use LaTeX notation for all mathematical equations and formulas. Enclose inline equations in $ symbols (like $x^2$) and display equations in $$ symbols (like $$\\frac{x}{y}$$)."
messages[0]["content"] = latex_instruction + " " + messages[0]["content"]

# Set stream=True to enable streaming and max_tokens to limit output
for chunk in ollama.chat.completions.create(
    model=model_name,
    messages=messages,
    stream=True,
    max_tokens=1500  # Set maximum token limit to 1500
):
    # Extract the content from the chunk
    content = chunk.choices[0].delta.content
    if content:
        print(content, end="")  # Print without newline
        full_response += content  # Build the complete response

# Display the full formatted response when complete
print("\n\nComplete response with rendered equations:\n")
display(Markdown(full_response))

# Ensure LaTeX is properly loaded in the notebook
from IPython.display import display, Math, Latex

# Show a sample equation to verify LaTeX rendering is working
display(Math(r'f(x) = x^2 + \frac{1}{x}'))
```
